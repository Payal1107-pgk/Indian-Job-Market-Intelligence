```python
import pandas as pd
import numpy as np

print("\n===== LOADING DATASET =====")

df = pd.read_excel("../data/indian-job-market-dataset-2025.xlsx")

print("Original dataset loaded!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df_clean = df.copy()

print("\nCleaning copy created successfully!")
```

```python
print("\n===== REMOVING EMPTY ROWS AND COLUMNS =====")

before_rows = df_clean.shape[0]
before_columns = df_clean.shape[1]

df_clean = df_clean.dropna(how="all")
df_clean = df_clean.dropna(axis=1, how="all")

print("Rows before:", before_rows)
print("Rows after:", df_clean.shape[0])

print("Columns before:", before_columns)
print("Columns after:", df_clean.shape[1])
```

```python
print("\n===== CLEANING COLUMN NAMES =====")

df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.replace(" ", "_")
)

print("Updated column names:")
print(df_clean.columns.tolist())
```

```python
print("\n===== CLEANING TEXT SPACES =====")

text_columns = df_clean.select_dtypes(include="object").columns

for column in text_columns:
    df_clean[column] = df_clean[column].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("Text spaces cleaned successfully!")
```

```python
print("\n===== STANDARDIZING MISSING VALUES =====")

missing_labels = [
    "Not disclosed",
    "Not Disclosed",
    "not disclosed",
    "N/A",
    "n/a",
    "NA",
    "na",
    "None",
    "none",
    "-"
]

df_clean = df_clean.replace(missing_labels, np.nan)

print("Missing-value labels standardized!")
```

```python
print("\n===== MISSING VALUES AFTER STANDARDIZATION =====")

missing_values = df_clean.isnull().sum()

print(missing_values[missing_values > 0])
```

```python
print("\n===== DUPLICATE CHECK =====")

duplicate_count = df_clean.duplicated().sum()

print("Duplicate rows found:", duplicate_count)
```

```python
print("\n===== REMOVING DUPLICATES =====")

before = len(df_clean)

df_clean = df_clean.drop_duplicates()

after = len(df_clean)

print("Rows before:", before)
print("Rows after:", after)
print("Duplicates removed:", before - after)
```

```python
print("\n===== CLEANING JOB TITLES =====")

df_clean["title"] = (
    df_clean["title"]
    .astype("string")
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

print("Job titles cleaned successfully!")
print("\nSample job titles:")
print(df_clean["title"].head(10))
```

```python
print("\n===== CLEANING COMPANY NAMES =====")

df_clean["companyName"] = (
    df_clean["companyName"]
    .astype("string")
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

print("Company names cleaned successfully!")
print("\nSample company names:")
print(df_clean["companyName"].head(10))
```

```python
print("\n===== CLEANING LOCATIONS =====")

df_clean["location"] = (
    df_clean["location"]
    .astype("string")
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

print("Locations cleaned successfully!")
print("\nSample locations:")
print(df_clean["location"].head(10))
```

```python
print("\n===== CLEANING SKILLS =====")

df_clean["tagsAndSkills"] = (
    df_clean["tagsAndSkills"]
    .astype("string")
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

print("Skills cleaned successfully!")
print("\nSample skills:")
print(df_clean["tagsAndSkills"].head(10))
```

```python
print("\n===== SALARY DATA INSPECTION =====")

print("\nSalary examples:")
print(df_clean["salary"].dropna().head(20))

print("\nMinimum salary examples:")
print(df_clean["minimumSalary"].dropna().head(20))

print("\nMaximum salary examples:")
print(df_clean["maximumSalary"].dropna().head(20))
```

```python
print("\n===== CHECKING ZERO SALARIES =====")

zero_min = (df_clean["minimumSalary"] == 0).sum()
zero_max = (df_clean["maximumSalary"] == 0).sum()

print("Minimum salary values equal to 0:", zero_min)
print("Maximum salary values equal to 0:", zero_max)
```

```python
print("\n===== SAMPLE ZERO-SALARY RECORDS =====")

zero_salary_records = df_clean[
    (df_clean["minimumSalary"] == 0) |
    (df_clean["maximumSalary"] == 0)
]

print(
    zero_salary_records[
        ["salary", "minimumSalary", "maximumSalary"]
    ].head(20).to_string(index=False)
)
```

```python
print("\n===== CHECKING SALARY CONSISTENCY =====")

invalid_salary = df_clean[
    df_clean["minimumSalary"] > df_clean["maximumSalary"]
]

print("Records where minimum salary > maximum salary:",
      len(invalid_salary))

print("\nSample invalid records:")

print(
    invalid_salary[
        ["salary", "minimumSalary", "maximumSalary"]
    ].head(10).to_string(index=False)
)
```

```python
print("\n===== CREATING AVERAGE SALARY =====")

df_clean["averageSalary"] = (
    df_clean["minimumSalary"] +
    df_clean["maximumSalary"]
) / 2

print("Average salary column created successfully!")

print("\nSample salary analysis:")

print(
    df_clean[
        ["salary", "minimumSalary", "maximumSalary", "averageSalary"]
    ].head(20).to_string(index=False)
)
```

```python
print("\n===== FIXING ZERO SALARY VALUES =====")

zero_min = (df_clean["minimumSalary"] == 0).sum()
zero_max = (df_clean["maximumSalary"] == 0).sum()

print("Zero minimum salaries:", zero_min)
print("Zero maximum salaries:", zero_max)

df_clean["minimumSalary"] = df_clean["minimumSalary"].replace(0, np.nan)
df_clean["maximumSalary"] = df_clean["maximumSalary"].replace(0, np.nan)

print("\nZero salary values converted to missing values.")
```

```python
print("\n===== EXPERIENCE CHECK =====")

print("\nExperience examples:")
print(
    df_clean["experience"]
    .dropna()
    .head(20)
    .to_string(index=False)
)

print("\nMinimum experience examples:")
print(
    df_clean["minimumExperience"]
    .dropna()
    .head(20)
    .to_string(index=False)
)

print("\nMaximum experience examples:")
print(
    df_clean["maximumExperience"]
    .dropna()
    .head(20)
    .to_string(index=False)
)
```

```python
print("\n===== CREATING AVERAGE EXPERIENCE =====")

df_clean["averageExperience"] = (
    df_clean["minimumExperience"] +
    df_clean["maximumExperience"]
) / 2

print("Average experience created successfully!")

print(
    df_clean[
        [
            "experience",
            "minimumExperience",
            "maximumExperience",
            "averageExperience"
        ]
    ].head(20).to_string(index=False)
)
```

```python
print("\n===== SKILLS DATA INSPECTION =====")

skills_data = df_clean["tagsAndSkills"].dropna()

print("Total job records with skills:", len(skills_data))

print("\nSample skill records:")

print(
    skills_data
    .head(20)
    .to_string(index=False)
)


print("\n===== MISSING SKILLS CHECK =====")

missing_skills = df_clean["tagsAndSkills"].isnull().sum()

print("Jobs without skill information:", missing_skills)

print(
    "Jobs with skill information:",
    df_clean["tagsAndSkills"].notnull().sum()
)
```

```python
print("\n===== MOST COMMON SKILL ENTRIES =====")

print(
    df_clean["tagsAndSkills"]
    .value_counts()
    .head(20)
    .to_string()
)


print("\n===== EXTRACTING INDIVIDUAL SKILLS =====")

skills_series = (
    df_clean["tagsAndSkills"]
    .dropna()
    .str.split(",")
    .explode()
    .str.strip()
)

print("Total skill entries:", len(skills_series))

print("\nSample individual skills:")

print(
    skills_series.head(30).to_string(index=False)
)

print("\n===== TOP 30 MOST DEMANDED SKILLS =====")

skill_counts = skills_series.value_counts()

print(
    skill_counts
    .head(30)
    .to_string()
)


print("\n===== CLEANING INDIVIDUAL SKILLS =====")

skills_series = skills_series[
    skills_series.notna() &
    (skills_series.str.strip() != "")
]

skills_series = skills_series.str.strip()

print("Valid skill entries:", len(skills_series))

print("\nTop 20 skills after cleaning:")

print(
    skills_series.value_counts()
    .head(20)
    .to_string()
)

print("\n===== CREATING SKILLS SUMMARY =====")

skill_summary = (
    skills_series
    .value_counts()
    .reset_index()
)

skill_summary.columns = [
    "Skill",
    "Job_Count"
]

print("Skills summary created successfully!")

print("\nTop 20 skills:")

print(
    skill_summary.head(20).to_string(index=False)
)


print("\n===== SAVING SKILLS SUMMARY =====")

skill_summary.to_csv(
    "../data/skill_demand_summary.csv",
    index=False
)

print("Skills summary saved successfully!")
```

```python
print("\n===== STANDARDIZING SKILL NAMES =====")

skills_series = skills_series.str.lower().str.strip()

print("Skill names standardized successfully!")

print("\n===== CREATING FINAL SKILL DEMAND TABLE =====")

skill_summary = (
    skills_series
    .value_counts()
    .reset_index()
)

skill_summary.columns = ["Skill", "Job_Count"]

print("Skill demand table created successfully!")

print("\nTop 20 demanded skills:")
print(skill_summary.head(20).to_string(index=False))

print("\n===== SAVING SKILL DEMAND TABLE =====")

skill_summary.to_csv(
    "../data/skill_demand_summary.csv",
    index=False
)

print("Skill demand table saved successfully!")
```

```python
print("\n===== TOP 20 JOB TITLES =====")

top_jobs = (
    df_clean["title"]
    .value_counts()
    .head(20)
)

print(top_jobs.to_string())
```

```python
print("\n===== TOP 20 HIRING COMPANIES =====")

top_companies = (
    df_clean["companyName"]
    .value_counts()
    .head(20)
)

print(top_companies.to_string())
```

```python
print("\n===== TOP 20 JOB LOCATIONS =====")

top_locations = (
    df_clean["location"]
    .value_counts()
    .head(20)
)

print(top_locations.to_string())

print("\n===== SALARY ANALYSIS =====")

salary_data = df_clean["averageSalary"].dropna()

print("Average Salary: ₹{:,.0f}".format(salary_data.mean()))

print("Median Salary: ₹{:,.0f}".format(salary_data.median()))

print("Minimum Salary: ₹{:,.0f}".format(salary_data.min()))

print("Maximum Salary: ₹{:,.0f}".format(salary_data.max()))

print("\n===== SALARY BY EXPERIENCE =====")

salary_experience = (
    df_clean[
        ["averageExperience", "averageSalary"]
    ]
    .dropna()
    .groupby("averageExperience")["averageSalary"]
    .mean()
    .sort_index()
)

print(salary_experience.to_string())

print("\n===== TOP 20 IN-DEMAND SKILLS =====")

print(
    skill_summary
    .head(20)
    .to_string(index=False)
)
```

```python
print("\n===== TOP 20 HIGHEST-PAYING LOCATIONS =====")

high_paying_locations = (
    df_clean[
        ["location", "averageSalary"]
    ]
    .dropna()
    .groupby("location")["averageSalary"]
    .mean()
    .sort_values(ascending=False)
    .head(20)
)

print(high_paying_locations.to_string())

print("\n===== SAVING FINAL CLEANED DATASET =====")

df_clean.to_csv(
    "../data/cleaned_job_market_data.csv",
    index=False
)

print("Final cleaned dataset saved successfully!")

print("\nRows:", len(df_clean))
print("Columns:", len(df_clean.columns))

```

```python
import pandas as pd

print("\n===== LOADING CLEANED DATASET =====")

df_clean = pd.read_csv("../data/cleaned_job_market_data.csv")

print("Cleaned dataset loaded successfully!")

print("\nRows:", len(df_clean))
print("Columns:", len(df_clean.columns))
```

```python
print("\n===== STORING CLEANED DATA IN SQLITE =====")

df_clean.to_sql(
    "job_market",
    conn,
    if_exists="replace",
    index=False
)

print("Table 'job_market' created successfully!")
print("Total records stored:", len(df_clean))
```

```python
print("\n===== VERIFYING JOB MARKET TABLE =====")

query = """
SELECT COUNT(*) AS total_jobs
FROM job_market;
"""

result = pd.read_sql_query(query, conn)

print(result.to_string(index=False))
```

```python
print("\n===== OVERALL JOB MARKET =====")

query = """
SELECT
    COUNT(*) AS total_jobs,
    COUNT(DISTINCT companyName) AS total_companies,
    COUNT(DISTINCT title) AS total_job_titles,
    COUNT(DISTINCT location) AS total_locations
FROM job_market;
"""

result = pd.read_sql_query(query, conn)

print(result.to_string(index=False))
```

```python
print("\n===== TOP 20 JOB TITLES =====")

query = """
SELECT
    title,
    COUNT(*) AS job_count
FROM job_market
WHERE title IS NOT NULL
GROUP BY title
ORDER BY job_count DESC
LIMIT 20;
"""

result = pd.read_sql_query(query, conn)

print(result.to_string(index=False))
```

```python
print("\n===== TOP 20 HIRING COMPANIES =====")

query = """
SELECT
    companyName,
    COUNT(*) AS job_count
FROM job_market
WHERE companyName IS NOT NULL
GROUP BY companyName
ORDER BY job_count DESC
LIMIT 20;
"""

result = pd.read_sql_query(query, conn)

print(result.to_string(index=False))
```

```python
print("\n===== TOP 20 JOB LOCATIONS =====")

query = """
SELECT
    location,
    COUNT(*) AS job_count
FROM job_market
WHERE location IS NOT NULL
GROUP BY location
ORDER BY job_count DESC
LIMIT 20;
"""

result = pd.read_sql_query(query, conn)

print(result.to_string(index=False))
```

```python
print("\n===== SALARY ANALYSIS =====")

query = """
SELECT
    ROUND(AVG(averageSalary), 0) AS average_salary,
    ROUND(MIN(averageSalary), 0) AS minimum_salary,
    ROUND(MAX(averageSalary), 0) AS maximum_salary
FROM job_market
WHERE averageSalary > 0;
"""

result = pd.read_sql_query(query, conn)

print(result.to_string(index=False))
```

```python
print("\n===== SALARY BY EXPERIENCE =====")

query = """
SELECT
    averageExperience,
    ROUND(AVG(averageSalary), 0) AS average_salary,
    COUNT(*) AS job_count
FROM job_market
WHERE averageExperience IS NOT NULL
  AND averageSalary > 0
GROUP BY averageExperience
ORDER BY averageExperience;
"""

result = pd.read_sql_query(query, conn)

print(result.to_string(index=False))
```

```python
print("\n===== HIGHEST-PAYING JOB TITLES =====")

query = """
SELECT
    title,
    ROUND(AVG(averageSalary), 0) AS average_salary,
    COUNT(*) AS job_count
FROM job_market
WHERE title IS NOT NULL
  AND averageSalary > 0
GROUP BY title
HAVING COUNT(*) >= 10
ORDER BY average_salary DESC
LIMIT 20;
"""

result = pd.read_sql_query(query, conn)

print(result.to_string(index=False))
```

```python
print("\n===== HIGHEST-PAYING LOCATIONS =====")

query = """
SELECT
    location,
    ROUND(AVG(averageSalary), 0) AS average_salary,
    COUNT(*) AS job_count
FROM job_market
WHERE location IS NOT NULL
  AND averageSalary > 0
GROUP BY location
HAVING COUNT(*) >= 10
ORDER BY average_salary DESC
LIMIT 20;
"""

result = pd.read_sql_query(query, conn)

print(result.to_string(index=False))
```

```python
print("\n===== CLOSING SQLITE CONNECTION =====")

conn.close()

print("SQLite connection closed successfully!")
```

```python
import pandas as pd

print("Pandas imported successfully!")


print("\n===== LOADING DATASET =====")

df = pd.read_excel("../data/indian-job-market-dataset-2025.xlsx")

print("Dataset loaded successfully!")


print("\n===== DATASET SIZE =====")

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])


print("\n===== DATASET INFORMATION =====")

df.info()

print("\n===== MISSING VALUES =====")

print(df.isnull().sum())

print("\n===== DUPLICATE RECORDS =====")

print("Number of duplicate rows:", df.duplicated().sum())

print("\n===== TOP 20 JOB TITLES =====")

print(df["title"].value_counts().head(20))

print("\n===== TOP 20 HIRING COMPANIES =====")

print(df["companyName"].value_counts().head(20))



print("\n===== TOP 20 JOB LOCATIONS =====")

print(df["location"].value_counts().head(20))

print("\n===== EXPERIENCE DISTRIBUTION =====")

print(df["experience"].value_counts())



print("\n===== SALARY INFORMATION =====")

print(df[["salary", "minimumSalary", "maximumSalary"]].head(20))


print("\n===== TOP SKILL ENTRIES =====")

print(df["tagsAndSkills"].value_counts().head(20))

print("\n===== JOB POSTING INFORMATION =====")

print(df["jobUploaded"].value_counts().head(20))


print("\n===== COMPANY RATINGS =====")

print(df["AggregateRating"].describe())

print("\n===== COMPANY REVIEWS =====")

print(df["ReviewsCount"].describe())

```
